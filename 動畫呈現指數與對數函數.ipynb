```python
import math
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np
from IPython.display import HTML

plt.rcParams['font.family'] = 'Microsoft JhengHei'

# 輸入
v0 = float(input("初速度 (m/s): "))
angle = float(input("角度 (度): "))

# 計算
theta = math.radians(angle)
g = 9.81
vx = v0 * math.cos(theta)
vy = v0 * math.sin(theta)

# 物理公式
t_max = vy / g  # 最高點時間
h_max = (vy**2) / (2*g)  # 最大高度
t_total = 2 * t_max  # 總時間
distance = vx * t_total  # 射程
print(f"總時間: {t_total:.1f} ")
print(f"最大高度: {h_max:.1f} m")
print(f"射程: {distance:.1f} m") 
print(f"飛行時間: {t_total:.2f} s")

# 建立軌跡點
t = np.linspace(0, t_total, 100) 
x = vx * t
y = vy * t - 0.5 * g * t**2

# 動畫
fig=Figure(figsize=(10, 6))#dpi解析度
ax=fig.add_subplot(111)

def animate(frame):
    ax.clear()
    ax.set_xlim(0, distance*1.1)  # 改成1.1倍就好
    ax.set_ylim(0, h_max*1.2)     # 改成1.2倍就好
    ax.set_title('拋物線運動')
    ax.grid(True)
    
    # 畫軌跡
    ax.plot(x[:frame+1], y[:frame+1], 'b-', linewidth=2)
    
    # 畫球
    if frame < len(x):
        ax.scatter(x[frame], y[frame], c='red', s=100)
        ax.text(16, 45, f't = {t[frame]:.2f} s')

anim = animation.FuncAnimation(fig, animate, frames=len(x), interval=50)
HTML(anim.to_jshtml())
```

```python
import math
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
from matplotlib.figure import Figure
plt.rcParams['font.family'] = 'Microsoft JhengHei'  # 顯示中文

# 物理參數
g = 9.8  # 重力加速度 (m/s²)
v0 = 0   # 初始速度 (m/s)
h0 = 100 # 初始高度 (m)

# 時間數據
time_data = []
height_data = []
velocity_data = []

# 計算 0 到 4.5 秒的數據（落地時間約 4.5 秒）
for i in range(46):  # 0.1秒間隔，共46個點
    t = i * 0.1
    
    # 自由落體公式: h = h0 + v0*t - (1/2)*g*t²
    h = h0 + v0*t - 0.5*g*t*t
    
    # 速度公式: v = v0 - g*t
    v = v0 - g*t
    
    if h >= 0:  # 只記錄還沒落地的數據
        time_data.append(t)
        height_data.append(round(h, 2))
        velocity_data.append(round(abs(v), 2))  # 取絕對值顯示速度大小
# 創建圖形

#fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
fig=Figure(figsize=(12, 6))#dpi解析度
ax1=fig.add_subplot(121)
ax2=fig.add_subplot(122)
def animate(frame):
    # 清除之前的圖
    ax1.clear()
    ax2.clear()
    
    # 取前 frame+1 個點
    current_time = time_data[:frame+1]
    current_height = height_data[:frame+1]
    current_velocity = velocity_data[:frame+1]
   
    # 左圖：高度-時間圖
    ax1.set_xlim(0, max(time_data))
    ax1.set_ylim(0, h0 + 5)
    ax1.set_title('自由落體 - 高度隨時間變化')
    ax1.set_xlabel('時間 (秒)')
    ax1.set_ylabel('高度 (公尺)')
    ax1.grid(True)
    
    # 繪製高度曲線
    ax1.plot(current_time, current_height, 'b-', linewidth=2, label='高度',marker='o')
     
    # 標示當前位置
    if current_height:
        ax1.scatter(current_time[-1], current_height[-1], c='red', s=100, marker='o')
        ax1.text(current_time[-1], current_height[-1] + 5, 
                f't={current_time[-1]:.1f}s\nh={current_height[-1]:.1f}m', 
                ha='center', fontsize=10, bbox=dict(facecolor="yellow"))
    
    ax1.legend()
    
    # 右圖：速度-時間圖
    ax2.set_xlim(0, max(time_data))
    ax2.set_ylim(0, max(velocity_data) + 5)
    ax2.set_title('自由落體 - 速度隨時間變化')
    ax2.set_xlabel('時間 (秒)')
    ax2.set_ylabel('速度 (公尺/秒)')
    ax2.grid(True)
    
    # 繪製速度曲線
    ax2.plot(current_time, current_velocity, 'r-', linewidth=2, label='速度',marker='o')
      
    # 標示當前速度
    if current_velocity:
        ax2.scatter(current_time[-1], current_velocity[-1], c='blue', s=100, marker='o')
        ax2.text(current_time[-1], current_velocity[-1] + 2, 
                f'v={current_velocity[-1]:.1f}m/s', 
                ha='center', fontsize=10, bbox=dict(facecolor="lightgreen"))
    
    ax2.legend()
    
    return ax1, ax2

# 建立動畫
anim = animation.FuncAnimation(fig, animate, frames=len(time_data), interval=200, repeat=True)

# 顯示動畫
plt.tight_layout()
HTML(anim.to_jshtml())
```

```python
#朝辭白帝彩雲間，千里江陵一日還
#李白搭船時剛好是汛期，順風時速約25公里、逆風約15公里，現代客輪每小時46公里
#白帝城到江陵水路約390公里
import math
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from matplotlib.figure import Figure
from IPython.display import HTML


plt.rcParams['font.family'] = ['Microsoft JhengHei']
plt.rcParams['axes.unicode_minus'] = False
# 模擬參數
distance = 390
times = []
fast_pos = []
slow_pos = []
now_pos=[]

# 計算數據
for i in range(63):  # 0.5秒間隔，24小時 120
    t = i * 0.5
    times.append(t)
    
    # 順風船 25km/h
    pos1 = min(25 * t, distance)
    fast_pos.append(pos1)
    
    # 逆風船 15km/h
    pos2 = min(15 * t, distance)
    slow_pos.append(pos2)

    #現代客輪
    pos3 = min(46 * t, distance) #46 92,......390
    now_pos.append(pos3)

# 創建圖形
fig = Figure(figsize=(12, 6))
ax = fig.add_subplot(111)

def animate(frame):
    # 清除之前的圖
    ax.clear()
    
    # 設定圖形範圍
    ax.set_xlim(0, 390)
    ax.set_ylim(-2, 2)
    ax.set_xlabel('距離 (公里)')
    ax.set_title('千里江陵真的能一日還?')
    
    # 當前時間和位置
    now = times[frame]
    p1 = fast_pos[frame]
    p2 = slow_pos[frame]
    p3=now_pos[frame]
    
    # 繪製船隻
    ax.plot(p1, 0.2, 'go', markersize=12, label='順風船 25km/h')
    ax.plot(p2, -0.2, 'ro', markersize=12, label='逆風船 15km/h')
    ax.plot(p3, 1.5, 'bo', markersize=12, label='現代客輪 46km/h')
    # 順風船時間標示
    if p1 >= distance:
        done_time = distance / 25
        ax.text(p1, 0.5, f'{done_time:.1f}小時', 
                ha='center', fontsize=10,
                bbox=dict(facecolor='lightgreen'))
    else:
        ax.text(p1, 0.5, f'{now:.1f}小時', 
                ha='center', fontsize=10,
                bbox=dict(facecolor='lightgreen'))
    
    # 逆風船時間標示
    if p2 >= distance:
        done_time = distance / 15
        ax.text(p2, -0.5, f'{done_time:.1f}小時', 
                ha='center', fontsize=10,
                bbox=dict(facecolor='lightyellow'))  # 修正這行
    else:
        ax.text(p2, -0.5, f'{now:.1f}小時', 
                ha='center', fontsize=10,
                bbox=dict(facecolor='lightyellow'))
    
    
    # 現代客輪時間標示
    if p3 >= distance:
        done_time = distance / 46
        ax.text(p3, 1.8, f'{done_time:.1f}小時', 
                ha='center', fontsize=10,
                bbox=dict(facecolor='lightblue'))  
    else:
        ax.text(p3, 1.8, f'{now:.1f}小時', 
                ha='center', fontsize=10,
                bbox=dict(facecolor='lightblue'))
    
    ax.legend()
    
    return ax,

# 建立動畫
anim = animation.FuncAnimation(fig, animate, frames=len(times), interval=200, repeat=True)

# 顯示動畫
HTML(anim.to_jshtml())
```
